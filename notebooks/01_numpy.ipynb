### Performance

```python
# Python Lists
arr1 = list(range(1000000))
arr2 = list(range(1000000, 2000000))

# Numpy Arrays
arr1_np = np.array(arr1)
arr2_np = np.array(arr2)
```

### Arithmatic operations and boradcasting

```python
arr1 = np.array([
  [1,2,3,4],
  [31,42,53,64],
  [51,62,73,84]
])

arr2 = np.array([
  [1,2,3,4],
  [31,42,53,64],
  [51,62,73,84]
])
```

```python
arr1+arr2
```

```python
arr1 * arr2
```

```python
# Broadcasting-এর জন্য দুটি dimension-এর যেকোনো একটি (1) হতে হবে অথবা দুটো সমান হতে হবে।

arr3 = np.array([
  [1,2,3,4]
])

arr4 = np.array([
  [1,2,3,4],
  [31,42,53,64],
  [51,62,73,84],
  [31,42,53,64]
])

arr3+arr4
```

```python
arr5 = np.array([
  [1,2,3,4],
  [3,4,5,6]
])

arr6 = np.array([
  [1,2,3,4],
  [3,4,5,6]
])
```

```python
display(arr5==arr6)
display((arr5==arr6).dtype)
```

```python
(arr5==arr6).sum
```
