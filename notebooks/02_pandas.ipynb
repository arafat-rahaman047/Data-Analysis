### Analyzing Data from data frames

```python
total_cases = covid_df.new_cases.sum()
total_deaths = covid_df.new_deaths.sum()

print('Reported Cases = {} & Reported Deaths = {}'.format(total_cases, total_deaths))
```

### Querying and sorting rows

```python
high_new_cases = covid_df.new_cases > 60000
high_new_cases
```

```python
covid_df[high_new_cases]
```

```python
high_cases_df = covid_df[covid_df.new_cases > 60000]
high_cases_df
```

```python
from IPython.display import display
with pd.option_context('display.max_rows',100):
  display(covid_df[covid_df.new_cases > 60000])
```

```python
positive_rate = (covid_df['new_cases'].sum()/covid_df['new_tests'].sum())
display(positive_rate)
```

```python
high_ratio_df = covid_df[covid_df.new_cases / covid_df.new_tests > positive_rate]
display(high_ratio_df)
```

```python
covid_df.new_cases / covid_df.new_tests
```

```python
covid_df['positive_rate'] = covid_df.new_cases / covid_df.new_tests
covid_df
```

```python
covid_df.drop(columns=['positive_rate'], inplace=True)
covid_df
```
